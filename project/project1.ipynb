```python
import io
import zipfile
import requests
import frontmatter
from collections import Counter
```

```python
url = 'https://codeload.github.com/modelcontextprotocol/modelcontextprotocol/zip/refs/heads/main'
resp = requests.get(url)
```

```python
repository_data = []

# Create a ZipFile object from the downloaded content
zf = zipfile.ZipFile(io.BytesIO(resp.content))

for file_info in zf.infolist():
    filename = file_info.filename.lower()

    # Only process markdown files (.md and .mdx)
    if not filename.endswith(('.md', '.mdx')):
        continue

    # Read and parse each file
    with zf.open(file_info) as f_in:
        content = f_in.read()
        post = frontmatter.loads(content)
        data = post.to_dict()
        data['filename'] = filename
        repository_data.append(data)

zf.close()
```

```python
print(repository_data[1])
```

```python
print("Documentos:", len(repository_data))
con_fm = [d for d in repository_data if len(d) > 2]  # más claves que content y filename
print("Con frontmatter:", len(con_fm))
print("Campos:", Counter(k for d in repository_data for k in d if k not in ("content", "filename")))
print("Extensiones:", Counter(d["filename"].rsplit(".", 1)[-1] for d in repository_data))
```
